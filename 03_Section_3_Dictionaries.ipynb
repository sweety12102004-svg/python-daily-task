## Q1. Check Student Result

````python
students = {"Rahul": 75, "Amit": 32, "Priya": 88, "Neha": 39, "Rohit": 55}

for name, marks in students.items():
    if marks >= 40:
        print(name, "Passed")
    else:
        print(name, "Failed")
````

**Expected output**

```text
Rahul Passed\nAmit Failed\nPriya Passed\nNeha Failed\nRohit Passed
```

## Q2. Student Performance Using Nested If

````python
students = {"Rahul": 85, "Amit": 35, "Priya": 72, "Neha": 48}

for name, marks in students.items():
    if marks >= 80:
        result = "Excellent"
    elif marks >= 60:
        result = "Good"
    elif marks >= 40:
        result = "Average"
    else:
        result = "Fail"
    print(name, result)
````

**Expected output**

```text
Rahul Excellent\nAmit Fail\nPriya Good\nNeha Average
```

## Q3. Product Stock Management

````python
products = {"Laptop": 10, "Mouse": 3, "Keyboard": 0, "Monitor": 7, "Printer": 2}

for name, stock in products.items():
    if stock == 0:
        status = "Out of Stock"
    elif stock <= 5:
        status = "Low Stock"
    else:
        status = "Available"
    print(name, status)
````

**Expected output**

```text
Laptop Available\nMouse Low Stock\nKeyboard Out of Stock\nMonitor Available\nPrinter Low Stock
```

## Q4. Employee Salary and Department

````python
employees = {
    "Rahul": {"salary": 60000, "department": "IT"},
    "Amit": {"salary": 40000, "department": "HR"},
    "Priya": {"salary": 70000, "department": "IT"},
    "Neha": {"salary": 45000, "department": "Finance"}
}

for name, info in employees.items():
    salary = info["salary"]
    department = info["department"]
    if salary >= 50000:
        if department == "IT":
            category = "Senior IT Employee"
        elif department == "HR":
            category = "Senior HR Employee"
        else:
            category = "Senior Employee"
    else:
        category = "Junior Employee"
    print(name, category)
````

**Expected output**

```text
Rahul Senior IT Employee\nAmit Junior Employee\nPriya Senior IT Employee\nNeha Junior Employee
```

## Q5. Check Even/Odd Values

````python
data = {"a": 10, "b": 15, "c": 25, "d": 30, "e": 40}

for key, value in data.items():
    if value % 2 == 0:
        if value > 20:
            result = "Large Even"
        else:
            result = "Even"
    else:
        result = "Odd"
    print(key, result)
````

**Expected output**

```text
a Even\nb Odd\nc Odd\nd Large Even\ne Large Even
```

## Q6. Shopping Cart Classification

````python
products = {"Mobile": 15000, "Mouse": 500, "Laptop": 60000, "Keyboard": 1200}

for name, price in products.items():
    if price >= 5000:
        category = "Premium Product"
    elif price >= 1000:
        category = "Expensive Product"
    else:
        category = "Affordable Product"
    print(name, category)
````


## Q7. Student Eligibility

````python
students = {
    "Rahul": {"attendance": 85, "marks": 70},
    "Amit": {"attendance": 60, "marks": 80},
    "Priya": {"attendance": 90, "marks": 35},
    "Neha": {"attendance": 78, "marks": 55}
}

for name, info in students.items():
    attendance = info["attendance"]
    marks = info["marks"]
    if attendance >= 75 and marks >= 40:
        result = "Eligible"
    elif attendance >= 75 and marks < 40:
        result = "Failed in Exam"
    else:
        result = "Not Eligible"
    print(name, result)
````

**Expected output**

```text
Rahul Eligible\nAmit Not Eligible\nPriya Failed in Exam\nNeha Eligible
```

## Q8. Bank Account Classification

````python
accounts = {"Rahul": 25000, "Amit": 5000, "Priya": 0, "Neha": -2000}

for name, balance in accounts.items():
    if balance >= 10000:
        status = "Premium Account"
    elif balance > 0:
        status = "Regular Account"
    elif balance == 0:
        status = "No Balance"
    else:
        status = "Overdraft"
    print(name, status)
````

**Expected output**

```text
Rahul Premium Account\nAmit Regular Account\nPriya No Balance\nNeha Overdraft
```

## Q9. Nested Dictionary - Employee Experience

````python
# The PDF provides rules but no data for this question; this is an illustrative dataset.
employees = {
    "Rahul": {"salary": 60000, "experience": 6},
    "Amit": {"salary": 45000, "experience": 4},
    "Priya": {"salary": 70000, "experience": 3},
    "Neha": {"salary": 30000, "experience": 1}
}

for name, info in employees.items():
    salary = info["salary"]
    exp = info["experience"]
    if salary >= 50000:
        category = "Senior" if exp >= 5 else "Mid-Level"
    else:
        category = "Experienced Junior" if exp >= 3 else "Fresher/Junior"
    print(name, category)
````

**Expected output**

```text
Rahul Senior\nAmit Experienced Junior\nPriya Mid-Level\nNeha Fresher/Junior
```

## Q10. Student Marks and Attendance

````python
students = {
    "Rahul": {"marks": 85, "attendance": 90},
    "Amit": {"marks": 35, "attendance": 80},
    "Priya": {"marks": 65, "attendance": 70},
    "Neha": {"marks": 45, "attendance": 85},
    "Rohit": {"marks": 90, "attendance": 60}
}

for name, info in students.items():
    marks = info["marks"]
    attendance = info["attendance"]
    if attendance < 75:
        result = "Not Eligible"
    elif marks >= 80:
        result = "Distinction"
    elif marks >= 60:
        result = "First Division"
    elif marks >= 40:
        result = "Pass"
    else:
        result = "Fail"
    print(name, result)
````

**Expected output**

```text
Rahul Distinction\nAmit Fail\nPriya Not Eligible\nNeha Pass\nRohit Not Eligible
```
