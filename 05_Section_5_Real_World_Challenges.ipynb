## Q1. ATM Withdrawal - Advanced Challenge

````python
pin = 1234
entered_pin = 1234
active = True
balance = 30000
amount = 25000

if entered_pin != pin:
    print("Invalid PIN")
elif not active:
    print("Account Inactive")
elif amount <= 0:
    print("Invalid Amount")
elif amount % 100 != 0:
    print("Amount must be a multiple of 100")
elif balance - amount < 500:
    print("Minimum balance of 500 required")
else:
    if amount > 20000:
        print("Extra Verification Required")
    balance -= amount
    print("Withdrawal Successful")
    print("Remaining Balance =", balance)
````

**Expected output**

```text
Extra Verification Required\nWithdrawal Successful\nRemaining Balance = 5000
```

## Q2. Electricity Bill

````python
units = 350

# The PDF does not clarify slab-wise versus flat-rate calculation.
# This example applies the rate for the total-unit range to every unit.
if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = units * 7
elif units <= 300:
    bill = units * 10
else:
    bill = units * 15

if bill > 3000:
    bill *= 1.10

print("Total Electricity Bill =", bill)
````

**Expected output**

```text
Total Electricity Bill = 5775.0
```

## Q3. Employee Bonus

````python
experience = 2
salary = 60000
performance = 95

if experience >= 1 and salary >= 50000:
    if performance >= 90:
        bonus_percent = 20
    elif performance >= 75:
        bonus_percent = 10
    else:
        bonus_percent = 5
elif salary < 50000:
    bonus_percent = 15 if performance >= 90 else 5
else:
    bonus_percent = 0

bonus = salary * bonus_percent / 100
print("Bonus Percentage =", bonus_percent)
print("Bonus Amount =", bonus)
````

**Expected output**

```text
Bonus Percentage = 20\nBonus Amount = 12000.0
```

## Q4. Driving License Eligibility

````python
age = 20
learner_license = True
test_score = 75

if age >= 18:
    if learner_license:
        if test_score >= 60:
            print("Eligible for Driving License")
        else:
            print("Driving Test Failed")
    else:
        print("Learner License Required")
else:
    print("Underage")
````

**Expected output**

```text
Eligible for Driving License
```
