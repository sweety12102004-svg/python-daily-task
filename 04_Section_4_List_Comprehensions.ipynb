## Q1. Find numbers greater than 50

````python
numbers = [20, 55, 30, 80, 45, 90, 10]
result = [n for n in numbers if n > 50]
print(result)
````

**Expected output**

```text
[55, 80, 90]
```

## Q2. Find numbers between 20 and 50 (inclusive)

````python
numbers = [10, 25, 35, 60, 45, 15, 50, 70]
result = [n for n in numbers if 20 <= n <= 50]
print(result)
````

**Expected output**

```text
[25, 35, 45, 50]
```

## Q3. Numbers divisible by both 3 and 5

````python
numbers = [10, 15, 20, 30, 45, 50, 60, 75]
result = [n for n in numbers if n % 3 == 0 and n % 5 == 0]
print(result)
````

**Expected output**

```text
[15, 30, 45, 60, 75]
```

## Q4. Names having more than 4 characters

````python
names = ["Amit", "Rahul", "Neha", "Rohit", "Sneha"]
result = [name for name in names if len(name) > 4]
print(result)
````

**Expected output**

```text
['Rahul', 'Rohit', 'Sneha']
```

## Q6. Extract vowels from a string

````python
text = "programming"
result = [ch for ch in text if ch in "aeiou"]
print(result)
````

**Expected output**

```text

```

## Q7. Extract digits from a string

````python
data = "Python123Programming456"
result = [ch for ch in data if ch.isdigit()]
print(result)
````

**Expected output**

```text
['1', '2', '3', '4', '5', '6']
```

## Q8. Replace numbers with Even/Odd labels

````python
numbers = [10, 15, 20, 25, 30]
result = ["Even" if n % 2 == 0 else "Odd" for n in numbers]
print(result)
````

**Expected output**

```text
['Even', 'Odd', 'Even', 'Odd', 'Even']
```

## Q9. Replace negative numbers with 0

````python
numbers = [10, -5, 20, -8, 30, -2]
result = [n if n >= 0 else 0 for n in numbers]
print(result)
````

**Expected output**

```text
[10, 0, 20, 0, 30, 0]
```

## Q10. Apply 10% discount for prices >= 1000, else 5%

````python
prices = [500, 1200, 800, 2000, 1500]
result = [p * 0.90 if p >= 1000 else p * 0.95 for p in prices]
print(result)
````

**Expected output**

```text
[475.0, 1080.0, 760.0, 1800.0, 1350.0]
```

## Q14. Separate deposits and withdrawals

````python
transactions = [5000, -2000, 3000, -1000, 7000, -500]
deposits = [t for t in transactions if t > 0]
withdrawals = [t for t in transactions if t < 0]
print("Deposits:", deposits)
print("Withdrawals:", withdrawals)
````

**Expected output**

```text
Deposits: [5000, 3000, 7000]\nWithdrawals: [-2000, -1000, -500]
```

## Q17. IT employees earning more than 60000

````python
employees = [
    {"name": "Amit", "department": "IT", "salary": 55000},
    {"name": "Rahul", "department": "HR", "salary": 70000},
    {"name": "Neha", "department": "IT", "salary": 80000},
    {"name": "Rohit", "department": "IT", "salary": 45000}
]
result = [
    e["name"] for e in employees
    if e["salary"] > 60000 and e["department"] == "IT"
]
print(result)
````

**Expected output**

```text
['Neha']
```
